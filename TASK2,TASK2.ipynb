```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': [],
    'C': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    print("BFS Traversal:")

    while queue:
        node = queue.popleft()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)

bfs(graph, 'A')
```

```python


heights = [1, 3, 5, 7, 6, 8, 9, 7, 5, 10]


current = 0

print("Starting position:", current)
print("Starting height:", heights[current])

while True:


    left = current - 1
    right = current + 1


    best = current


    if left >= 0 and heights[left] > heights[best]:
        best = left


    if right < len(heights) and heights[right] > heights[best]:
        best = right

    if best == current:
        print("\nNo better neighbor found!")
        print("Hill climbing stopped.")
        break


    current = best

    print("Moved to position:", current,
          "Height:", heights[current])

print("\nFinal position:", current)
print("Final height:", heights[current])
```

```python
from collections import deque

def find_path(rows, cols, start, end):
    queue = deque([start])
    visited = {start}
    parent = {start: None}

    directions = [
        (1, 0),   # down
        (-1, 0),  # up
        (0, 1),   # right
        (0, -1)   # left
    ]

    while queue:
        current = queue.popleft()

        if current == end:
            break

        r, c = current

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            next_cell = (nr, nc)

            if 0 <= nr < rows and 0 <= nc < cols:
                if next_cell not in visited:
                    visited.add(next_cell)
                    parent[next_cell] = current
                    queue.append(next_cell)

    # Reconstruct path
    if end not in parent:
        return []

    path = []
    current = end

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()
    return path


path = find_path(3, 3, (0, 0), (1, 2))

print("Path:", path)
print("Path length:", len(path) - 1)
```

```python
def find_all_paths(rows, cols, start, end):
    directions = [
        (1, 0),   # down
        (-1, 0),  # up
        (0, 1),   # right
        (0, -1)   # left
    ]

    all_paths = []

    def dfs(current, path, visited):
        if current == end:
            all_paths.append(path.copy())
            return

        r, c = current

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            next_cell = (nr, nc)

            if (0 <= nr < rows and
                0 <= nc < cols and
                next_cell not in visited):

                visited.add(next_cell)
                path.append(next_cell)

                dfs(next_cell, path, visited)

                # Backtrack
                path.pop()
                visited.remove(next_cell)

    dfs(start, [start], {start})

    return all_paths


# Find all paths
paths = find_all_paths(3, 3, (0, 0), (1, 2))

# Print paths and their lengths
for i, path in enumerate(paths, 1):
    print(f"Path {i}: {path}")
    print(f"Path length: {len(path) - 1}")
    print()
```

```python
import heapq

class Node:
    def __init__(self, position, parent=None):
        self.position = position
        self.parent = parent
        self.g = 0
        self.h = 0
        self.f = 0

    def __lt__(self, other):
        return self.f < other.f

def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])  # Manhattan distance

def astar(grid, start, goal):
    open_list = []
    closed_list = set()
    start_node = Node(start)
    goal_node = Node(goal)
    heapq.heappush(open_list, start_node)

    while open_list:
        current_node = heapq.heappop(open_list)
        closed_list.add(current_node.position)

        if current_node.position == goal_node.position:
            path = []
            while current_node:
                path.append(current_node.position)
                current_node = current_node.parent
            return path[::-1], len(path)-1  # path and cost

        neighbors = [(0,1),(0,-1),(1,0),(-1,0)]
        for dx, dy in neighbors:
            neighbor_pos = (current_node.position[0] + dx, current_node.position[1] + dy)

            if (0 <= neighbor_pos[0] < len(grid) and
                0 <= neighbor_pos[1] < len(grid[0]) and
                grid[neighbor_pos[0]][neighbor_pos[1]] == 0 and
                neighbor_pos not in closed_list):

                neighbor_node = Node(neighbor_pos, current_node)
                neighbor_node.g = current_node.g + 1
                neighbor_node.h = heuristic(neighbor_pos, goal_node.position)
                neighbor_node.f = neighbor_node.g + neighbor_node.h

                if not any(open_node.position == neighbor_node.position and open_node.g <= neighbor_node.g for open_node in open_list):
                    heapq.heappush(open_list, neighbor_node)

    return None, None

# 3x3 grid (0 = free, 1 = obstacle)
grid = [
    [0,0,0],
    [0,1,0],
    [0,0,0]
]

start = (0,0)
goal = (2,2)

path, cost = astar(grid, start, goal)
print("Path:", path)
print("Cost:", cost)
```

```python
import heapq

def heuristic(a, b):

    return abs(a[0] - b[0]) + abs(a[1] - b[1])


def a_star(rows, cols, start, goal):

    open_list = []
    heapq.heappush(open_list, (0, 0, start, [start]))

    g_cost = {start: 0}

    directions = [
        (0, 1),
        (1, 0),
        (0, -1),
        (-1, 0)
    ]

    while open_list:
        f, g, current, path = heapq.heappop(open_list)

        if current == goal:
            return path, g

        for dr, dc in directions:
            nr = current[0] + dr
            nc = current[1] + dc
            neighbor = (nr, nc)

            if 0 <= nr < rows and 0 <= nc < cols:

                new_g = g + 1


                if neighbor not in g_cost or new_g < g_cost[neighbor]:
                    g_cost[neighbor] = new_g

                    h = heuristic(neighbor, goal)
                    f = new_g + h

                    heapq.heappush(
                        open_list,
                        (f, new_g, neighbor, path + [neighbor])
                    )

    return None, None



rows = int(input("enter number of row:"))
cols = int(input("enter number of coloumn:"))

start = (0, 0)
goal = (rows - 1, cols - 1)

print("start", start)
print("goal =", goal)


path, cost = a_star(rows, cols, start, goal)

if path:
    print("\noptimal path:", end="")
    print(path)
    print("cost :", cost)
else:
    print("No path found")

```

```python
def minimax(values, depth, index, maximizing_player):
    # If we reach a leaf node
    if depth == 0:
        return values[index]

    # Maximizing player
    if maximizing_player:
        return max(
            minimax(values, depth - 1, index * 2, False),
            minimax(values, depth - 1, index * 2 + 1, False)
        )

    # Minimizing player
    else:
        return min(
            minimax(values, depth - 1, index * 2, True),
            minimax(values, depth - 1, index * 2 + 1, True)
        )


# Input
height = int(input("enter tree height:"))

n = 2 ** height
print("enter", n, "leaf node values:")

values = list(map(int, input().split()))

# Minimax calculation
result = minimax(values, height, 0, True)

print("leaf values :", values)
print("min max value :", result)
```
